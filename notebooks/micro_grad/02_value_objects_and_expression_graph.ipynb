## 5. Create an Addition Value Node

Create a scalar computation-graph node for addition. Each of left and right is a Python dictionary with exactly two fields: id, a string identifier, and data, its numeric value.

$$ z = x + y $$

where,
y are the left and right data values, and z is the new node's data. This task constructs the forward node only; no backward pass is required.
Return a new Python dictionary with exactly these fields: id containing output_id, data containing the sum as a Python float, grad containing 0.0 as a Python float, op containing "+", and parents containing a two-item list of the original left and right dictionaries in that order.

```python
def value_addition_node(left: dict, right: dict, output_id: str) -> dict:
    """
    Returns an addition node with id, data, grad, op, and the original parents.
    """
    output_dict = {}
    output_dict['id'] = output_id
    output_dict['data'] = float(left['data']) + float(right['data'])
    output_dict['grad'] = float(0.0)
    output_dict['op'] = '+'
    output_dict['parents'] = [left,right]
    return output_dict
    
```

## 6. Create a Multiplication Value Node

Create a scalar multiplication node from the left and right leaf dictionaries, each containing id and data.

$$ z = x * y $$
Here x and y are the left and right data values, and z is their product.

Return a new dictionary with exactly id, data, grad, op, and parents: id is output_id, data is z as a Python float, grad is 0.0, op is the string "*", and parents is a list containing the original left and right objects in that order.

```python
def value_multiplication_node(left: dict, right: dict, output_id: str) -> dict:
    """
    Returns the node dictionary with id, data, grad, op, and parents.
    """
    output_dict = {}
    output_dict['id'] = output_id
    output_dict['data'] = float(left['data']) * float(right['data'])
    output_dict['grad'] = float(0.0)
    output_dict['op'] = '*'
    output_dict['parents'] = [left,right]
    return output_dict
    
```
